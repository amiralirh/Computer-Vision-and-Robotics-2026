```python
import numpy as np
```

```python
def incompetence_score(transforms):
    s = 0

    R_t = None
    for R_deltat in transforms:
        if R_t is None:
            R_t = R_deltat
            continue

        Rrel = R_t.T @ R_deltat
        R_t = R_deltat
        cos_theta = np.clip((np.trace(Rrel) - 1) / 2, -1.0, 1.0)
        theta = np.arccos(cos_theta)
        theta = np.rad2deg(theta)
        s += theta**2

    return s

```

```python
A = np.array([[0.9998, -0.0175, 0],
              [0.0175, 0.9998, 0],
               [0,0,1]])

B = np.array([[0.9962, -0.0174, 0.0858],
              [0.0175, 0.9998, 0],
               [-0.0858,0.0015,0.9963]])



transforms = np.array([np.eye(3), A, B])
s = incompetence_score(transforms)
print(f'Incompetence Score = {s:.3F}')
```
